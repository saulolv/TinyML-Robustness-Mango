```python
import json
import os

def get_file_size_mb(path):
    return round(os.path.getsize(path) / (1024 * 1024), 2)

model_files = {
    "efficientnetb0": {
        "zip":   "efficientnetb0.zip",
        "tflite": "efficientnetb0.tflite",
        "keras": "efficientnetb0.keras"
    },
    "mobilenetv3": {
        "zip":   "mobilenetv3_small.zip",
        "tflite": "mobilenetv3_small.tflite",
        "keras": "mobilenetv3_small.keras"
    },
    "lcnn": {
        "zip":   "lcnn.zip",
        "tflite": "lcnn.tflite",
        "keras": "lcnn.keras"
    }
}

sizes = {}

for model_name, files in model_files.items():
    sizes[model_name] = {}
    for fmt, fname in files.items():
        if os.path.exists(fname):
            sizes[model_name][fmt] = get_file_size_mb(fname)
        else:
            sizes[model_name][fmt] = None

with open("model_sizes.json", "w") as f:
    json.dump(sizes, f, indent=4)

print("Saved model sizes as model_sizes.json")
```
